```python
import string
from collections import Counter
from nltk.tokenize import TweetTokenizer

# This inputs in the file that will be tokenized and counted
input_file_path = "test01_cc_sharealike.txt"

# This creates a variable initializing the Tweet Tokenizer
tknzr = TweetTokenizer()

with open(input_file_path, "r", encoding="utf-8") as f:
    text = f.read().lower()

tokens = tknzr.tokenize(text)

# This line works to filter (or skip) the punctuation in the tokenized file contents
word_tokens = [t for t in tokens if t not in string.punctuation]

# This code segment takes the counts and then collects the top 10 most commonly found words in the file
count = Counter(word_tokens)
top_10 = count.most_common(10)

# Finally, this displays the most common words and their counts in an output tab
for word, num in count.most_common(10):
    print(word, "\t", num)
```
